```python
import os
import numpy as np
```

```python
def compute_average_bbox(openeds_root):
    bbox_dir = os.path.join(openeds_root, 'bbox_preds')
    xmin_list, xmax_list, ymin_list, ymax_list = [], [], [], []

    for root, _, files in os.walk(bbox_dir):
        for file in files:
            if file.endswith(".txt"):
                file_path = os.path.join(root, file)
                with open(file_path, 'r') as f:
                    for line in f:
                        parts = line.strip().split(',')
                        if len(parts) == 4:
                            try:
                                xmin, xmax, ymin, ymax = map(float, parts)
                                xmin_list.append(xmin)
                                xmax_list.append(xmax)
                                ymin_list.append(ymin)
                                ymax_list.append(ymax)
                            except ValueError:
                                continue  # skip bad lines

    if not xmin_list:
        print("No valid bounding boxes found.")
        return

    print("Average xmin:", np.mean(xmin_list))
    print("Average xmax:", np.mean(xmax_list))
    print("Average ymin:", np.mean(ymin_list))
    print("Average ymax:", np.mean(ymax_list))
```
