```python
import matplotlib.pyplot as plt

# Suppose you have…
models   = [
            'SERTnetV1', 
            'SERTnetV2', 
            'SERTnetV3', 
            'SERTnetV4',
            'SERTnetV5',
            ]  # names of your models
internal = [
            0.8501180229920807,      
            0.8506252720700048,        
            0.82,    
            0.90
            ]         # validation‐set mIoUs
external = [
            np.mean([0.99185008, 0.91011085, 0.96545823, 0.95573928]),      
            0.84,        
            0.79,    
            0.88
            ]         # test‐set   mIoUs

fig, ax = plt.subplots(figsize=(6,6))
ax.scatter(internal, external, s=100)

# draw y=x line as reference
lims = [min(min(internal),min(external)) - 0.02,
        max(max(internal),max(external)) + 0.02]
ax.plot(lims, lims, '--', color='gray', linewidth=1)
ax.set_xlim(lims)
ax.set_ylim(lims)

# annotate each point with its model name
for i, name in enumerate(models):
    ax.annotate(name, (internal[i], external[i]),
                textcoords="offset points", xytext=(5,-5))

ax.set_xlabel('Internal validation mIoU')
ax.set_ylabel('External  test  mIoU')
ax.set_title('Internal vs. External Performance')
ax.grid(True)
fig.tight_layout()
plt.savefig('internal_vs_external.png', dpi=150)
plt.show()
```
