```python
import os
import numpy as np
from tqdm import tqdm
```

```python
def find_blank_masks(openeds_root):
    eyes_closed = []

    # Get all subject folders
    subjects = [d for d in os.listdir(openeds_root) if d.startswith('S_') and os.path.isdir(os.path.join(openeds_root, d))]

    for subject in tqdm(subjects, desc="Checking subjects"):
        subject_path = os.path.join(openeds_root, subject)
        mask_files = [f for f in os.listdir(subject_path) if f.endswith('.npy')]

        for mask_file in mask_files:
            mask_path = os.path.join(subject_path, mask_file)
            try:
                mask = np.load(mask_path)
                if np.all(mask == 0):  # Check if mask is entirely blank
                    eyes_closed.append(f"{subject}/{mask_file}")
            except Exception as e:
                print(f"Failed to load {mask_path}: {e}")

    # Save the results to eyes_closed.txt
    output_file = 'eyes_closed.txt'
    with open(output_file, 'w') as f:
        for line in eyes_closed:
            f.write(f"{line}\n")

    print(f"Finished! Found {len(eyes_closed)} closed-eye masks.")
    print(f"Saved to {output_file}")
```
